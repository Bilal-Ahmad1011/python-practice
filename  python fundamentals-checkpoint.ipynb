# 2. Data Types

```python
# integer
print(5)
print(1e308)   # 1*10^308
print(1e309)   # this means python can handle only integers upto 1*10^308
```

```python
# Decimal/Float
print(3.32)
print(1.7e308)   # max float value 1.7*10^308
print(1.7e309)
```

```python
# Boolean
print(True)   # Boolean is not text in this we dont use inverted commas""
print(False) 
```

```python
# Text/String
print('fellow') 
```

```python
# complex
print(5+7j) 
```

```python
# List-> /Array
print([1,2,3,4,5])
```

```python
# Tuple
print((1,2,3,4))
```

```python
# Sets
print({1,2,3,4})
```

```python
# Dictionary
print({'name':'harry','gender':'male'})
```

```python
# type
type(5)
type('hello')
```

# 3. Variables

```python
name = 'Bilal'
print(name)
```

```python
a = 5
b = 7
print(a+b)
```

```python
# Dynamic Typing
a = 5   # means there is no need of declaring variable name

# Static Typing 
int a = 5    # Here there is need of declaring variable name  

   # python uses Static Typing
```

```python
# Dynamic Binding
a = 5
print(a)
a = 'Bilal'
print(a)
  # in python there is no fixed data type of variable , in python a variable can hold multiple data types in a same program
```

```python
# static Binding
c/c++ , java supports this. means if a is used as integer it cant be used as string in next time
```

```python
# creating multiple variables
a = 3
b = 4
c = 5
print(a,b,c)
```

```python
a,b,c = 3,4,5
print(a,b,c)
```

```python
# Giving same value to multiple variables

a=b=c= 5
print(a,b,c)
```
